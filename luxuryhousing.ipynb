```python
# Create lowercase mapping
micro_market_mapping = {
    'jayanagar': 'Jayanagar',
    'bannerghatta road': 'Bannerghatta Road',
    'sarjapur road': 'Sarjapur Road',
    'jp nagar': 'JP Nagar',
    'whitefield': 'Whitefield',
    'hennur road': 'Hennur Road',
    'mg road': 'MG Road',
    'indiranagar': 'Indiranagar',
    'yelahanka': 'Yelahanka',
    'kanakapura road': 'Kanakapura Road',
    'bellary road': 'Bellary Road',
    'electronic city': 'Electronic City',
    'domlur': 'Domlur',
    'koramangala': 'Koramangala',
    'hebbal': 'Hebbal',
    'rajajinagar': 'Rajajinagar'
}

# Normalize and map
df['Micro_Market'] = (
    df['Micro_Market']
    .str.strip()                  # remove spaces
    .str.lower()                  # make lowercase
    .map(micro_market_mapping)    # map to clean names
)

# Optional: check for unmapped values
unmapped = df[df['Micro_Market'].isna()]
print(unmapped['Micro_Market'].unique())
```
